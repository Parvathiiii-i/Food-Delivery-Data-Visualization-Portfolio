```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# 1. Load Dataset
df = pd.read_csv("Day14_Food_Delivery_Visualization_Dataset.csv")
df["Date"] = pd.to_datetime(df["Date"])
sns.set_theme(style="whitegrid")

# 2. Line Plot: Trend of Revenue Over Time
plt.figure(figsize=(10, 4))
df_sorted = df.sort_values("Date")
plt.plot(
    df_sorted["Date"],
    df_sorted["Revenue"],
    color="b",
    marker="o",
    linestyle="-",
    linewidth=1,
    markersize=3,
)
plt.title("Trend of Revenue Over Time")
plt.xlabel("Date")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
print(
    "Interpretation: Revenue fluctuates regularly over time with seasonal spikes peaking towards late 2025."
)

# 3. Bar Chart: Total Revenue by City
plt.figure(figsize=(8, 4))
city_rev = df.groupby("City")["Revenue"].sum().reset_index()
sns.barplot(data=city_rev, x="City", y="Revenue", palette="viridis")
plt.title("Total Revenue by City")
plt.xlabel("City")
plt.ylabel("Total Revenue")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
print(
    "Interpretation: Bengaluru and Delhi lead in total revenue generation compared to other cities."
)

# 4. Scatter Plot: Marketing Spend vs Revenue
plt.figure(figsize=(7, 4))
sns.scatterplot(data=df, x="Marketing_Spend", y="Revenue", hue="City", alpha=0.8)
plt.title("Marketing Spend vs Revenue")
plt.xlabel("Marketing Spend")
plt.ylabel("Revenue")
plt.legend(bbox_to_anchor=(1.05, 1), loc="upper left")
plt.tight_layout()
plt.show()
print(
    "Interpretation: Shows a moderate positive relationship between marketing expenditure and resulting revenue."
)

# 5. Scatter Plot: Delivery Time vs Customer Rating
plt.figure(figsize=(7, 4))
sns.scatterplot(
    data=df, x="Avg_Delivery_Minutes", y="Customer_Rating", color="red", alpha=0.7
)
plt.title("Delivery Time vs Customer Rating")
plt.xlabel("Average Delivery Time (Minutes)")
plt.ylabel("Customer Rating")
plt.tight_layout()
plt.show()
print(
    "Interpretation: Strong negative correlation; as delivery time increases, customer ratings drop significantly."
)

# 6. Histogram: Distribution of Average Order Value
plt.figure(figsize=(7, 4))
sns.histplot(df["Average_Order_Value"], bins=20, kde=True, color="purple")
plt.title("Distribution of Average Order Value")
plt.xlabel("Average Order Value")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
print(
    "Interpretation: Average order value is multi-modal and distributed roughly between 250 and 550."
)

# 7. Box Plot: Delivery Time by Weather
plt.figure(figsize=(7, 4))
sns.boxplot(data=df, x="Weather", y="Avg_Delivery_Minutes", palette="Set2")
plt.title("Delivery Time Distribution Across Weather Conditions")
plt.xlabel("Weather")
plt.ylabel("Average Delivery Time (Minutes)")
plt.tight_layout()
plt.show()
print(
    "Interpretation: Rainy and hot weather conditions cause longer and more variable delivery times."
)

# 8. Violin Plot: Customer Rating by Order Channel
plt.figure(figsize=(7, 4))
sns.violinplot(data=df, x="Order_Channel", y="Customer_Rating", palette="Pastel1")
plt.title("Customer Rating Distribution by Order Channel")
plt.xlabel("Order Channel")
plt.ylabel("Customer Rating")
plt.tight_layout()
plt.show()
print(
    "Interpretation: App orders exhibit slightly higher median ratings compared to partner platforms and websites."
)

# 9. Count Plot: Orders by Weather
plt.figure(figsize=(7, 4))
sns.countplot(
    data=df, x="Weather", order=df["Weather"].value_counts().index, palette="muted"
)
plt.title("Count of Orders by Weather Condition")
plt.xlabel("Weather")
plt.ylabel("Count")
plt.tight_layout()
plt.show()
print(
    "Interpretation: Clear weather records the highest volume of orders, whereas extreme weather conditions see fewer orders."
)

# 10. Correlation Heatmap
plt.figure(figsize=(8, 6))
numeric_cols = df.select_dtypes(include=["float64", "int64"]).columns
corr = df[numeric_cols].corr()
sns.heatmap(
    corr, annot=True, fmt=".2f", cmap="coolwarm", linewidths=0.5, cbar=True
)
plt.title("Correlation Heatmap of Numerical Features")
plt.tight_layout()
plt.show()
print(
    "Interpretation: Highlights strong positive correlation between orders and revenue, and strong negative correlation between delivery time and customer ratings."
)
```
